```python
def ip_vers_entier(s):
    L=s.split('.')
    val=0
    for i in range(4):
        val+=int(L[i])*(256**(3-i)) 
    return val
s="192.168.1.1"
print(ip_vers_entier(s))
```

```python
def entier_vers_ip(x):
    ip=""
    while x>0:
        ip=str(x%256)+"."+ip
        x=x//256
    return ip[:-1]
x=3232235777
print(entier_vers_ip(x))
    
```

```python
def masque(n):
    return (2**n-1)<<(32-n)

entier_vers_ip(masque(24))
```

```python
def masque2(n):
    return int("1"*n+"0"*(32-n),2)

entier_vers_ip(masque2 (24))
```

```python
masque2(24) == 2**32 - 2**(32-24)
```

```python
def adresse_reseau(a, n):
    return entier_vers_ip(ip_vers_entier(a) & masque(n))
       
```

```python
a="192.168.1.123"
n=24
adresse_reseau(a, n)
```

```python
def  diffusion(a, n):
    return entier_vers_ip(ip_vers_entier(adresse_reseau(a, n))+int("1"*(32-n),2))
```

```python
N=0xFFFF
def add16(a, b) :
    return (a+b)%N +(a+b)//N
```

```python
def mots16(octets):
    mots=[]
    for i in range(0,len(octets),2):
        mots.append(octets[i]*256+octets[i+1])
    if len(octets)%2==1:
        mots.append(octets[-1]*256)
    return mots
```

```python
[ hex (p) for p in mots16([0x16,0x00,0x18,0x13])]
```

```python
def somme_un(mots):
    s=0
    for m in mots:
        s=add16(s,m)
    return s
```

```python
def checksum(octets):
    return somme_un(mots16([0x16,0x00,0x18,0x13]))
```
